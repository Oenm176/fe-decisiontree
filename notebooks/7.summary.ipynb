```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# Data Rekapitulasi (Berdasarkan hasil kamu)
results = {
    'Dataset': ['Cleveland (Jantung)', 'Cleveland (Jantung)', 'Glass (Kaca)', 'Glass (Kaca)'],
    'Metode': ['Data Mentah', 'Data Bersih (+3%)', 'Data Mentah', 'Data Bersih (+3%)'],
    'Akurasi': [80.3, 83.5, 76.7, 79.1]  # Masukkan angka persis hasilmu di sini
}

df_res = pd.DataFrame(results)

# Plotting
plt.figure(figsize=(10, 6))

# Membuat Bar Chart dengan warna beda
ax = sns.barplot(data=df_res, x='Dataset', y='Akurasi', hue='Metode', palette=['#ff9999', '#66b3ff'])

# Menambahkan Label Angka di atas batang
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', padding=3, fontsize=11, fontweight='bold')

# Dekorasi
plt.title('Konsistensi Peningkatan Akurasi (~3%)\nCleveland vs Glass Dataset', fontsize=14, fontweight='bold')
plt.ylim(0, 100)
plt.ylabel('Akurasi (%)')
plt.xlabel('')
plt.legend(title='Kondisi Data', loc='lower right')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

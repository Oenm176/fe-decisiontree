## Check Data Hilang

```python
check_null = dataset_1.isna().sum()
if check_null.sum() > 0:
    print(f"\nKolom dengan Missing Values dataset 1:")
    print(f"{check_null[check_null > 0]}")
else:
    print(f"Tidak ditemuka data kosong atau Null dataset 1")
```
